```python
import pandas as pd
from pathlib import Path

# Define o arquivo original
arquivo = Path("04_PENSE2024_SEM_NAN.parquet")

# Define o nome do novo arquivo no mesmo diretório
novo_arquivo = arquivo.with_name("05_PENSE2024_SEM_NAN_SEM_DUPLICADOS.parquet")

# Carrega o DataFrame original
df = pd.read_parquet(arquivo)

# Registra a quantidade original de linhas
linhas_antes = len(df)

# Remove linhas duplicadas, mantendo a primeira ocorrência
df_limpo = df.drop_duplicates(keep="first").copy()

# Garantir que todas as variáveis continuem como string após a remoção das duplicatas
for coluna in df_limpo.columns:
    df_limpo[coluna] = df_limpo[coluna].astype("string")

# Registra a quantidade final de linhas
linhas_depois = len(df_limpo)

# Calcula quantas linhas foram removidas
removidas = linhas_antes - linhas_depois

# Salva o novo arquivo Parquet
df_limpo.to_parquet(novo_arquivo, index=False)

# Exibe o resumo
print(f"Linhas antes da remoção: {linhas_antes:,}")
print(f"Linhas após a remoção: {linhas_depois:,}")
print(f"Linhas removidas: {removidas:,}")
print(f"\nNovo arquivo salvo em: {novo_arquivo.resolve()}")
```

```python
import pandas as pd
from IPython.display import HTML, display


# ============================================================
# 1. CARREGAR O DICIONÁRIO DE DADOS
# ============================================================

dicionario = pd.read_excel(
    "00_Dicionário_Microdados_PENSE2024.xlsx",
    engine="openpyxl"
)


# ============================================================
# 2. FUNÇÃO PARA NORMALIZAR TEXTOS
#
# Remove espaços e transforma tudo em minúsculas.
# Isso evita problemas de comparação entre textos.
# ============================================================

def normalizar(texto):
    return "".join(
        str(texto).split()
    ).lower()


# ============================================================
# 3. CRIAR O MAPEAMENTO DAS CATEGORIAS
#
# A chave será a "Descrição da Variável".
#
# Exemplo:
#
# "código de regiões"
#
# será associado a:
#
# 1 -> norte
# 2 -> nordeste
# 3 -> sudeste
# 4 -> centro-oeste
# 5 -> sul
# ============================================================

dico_map = {}


for _, row in dicionario.iterrows():

    # --------------------------------------------------------
    # Obtém a descrição da variável
    # --------------------------------------------------------

    desc = str(
        row["Descrição da Variável"]
    ).strip()


    # Ignora descrições vazias
    if not desc or desc == "nan":
        continue


    # Normaliza a descrição
    chave = normalizar(desc)


    # --------------------------------------------------------
    # Obtém as categorias
    # --------------------------------------------------------

    cat = (
        str(row["Categoria"])
        if pd.notna(row["Categoria"])
        else ""
    )


    cat_dict = {}


    # --------------------------------------------------------
    # Processa cada categoria
    # --------------------------------------------------------

    if cat:

        for linha in cat.split("\n"):

            linha = linha.strip()

            if " - " in linha:

                codigo, significado = linha.split(
                    " - ",
                    1
                )

                codigo = codigo.strip()
                significado = significado.strip()


                # Guarda o código exatamente como aparece
                cat_dict[codigo] = significado


                # ------------------------------------------------
                # Também aceita códigos numéricos em formatos
                # diferentes:
                #
                # 1
                # 1.0
                #
                # Isso é importante porque o Parquet pode
                # armazenar o mesmo valor de maneiras diferentes.
                # ------------------------------------------------

                try:

                    codigo_int = str(
                        int(float(codigo))
                    )

                    cat_dict[codigo_int] = significado

                except (ValueError, TypeError):

                    pass


                try:

                    codigo_float = str(
                        float(codigo)
                    )

                    cat_dict[codigo_float] = significado

                except (ValueError, TypeError):

                    pass


    # Guarda o mapa da variável
    dico_map[chave] = cat_dict


# ============================================================
# 4. GERAR A TABELA RESUMO
# ============================================================

tabela_dados = []


for col in df_limpo.columns:

    # --------------------------------------------------------
    # Nome completo da coluna
    #
    # Exemplo:
    #
    # REGIAO - código de regiões
    # --------------------------------------------------------

    col_limpa = str(
        col
    ).strip()


    # ========================================================
    # 4.1 SEPARAR CÓDIGO E DESCRIÇÃO
    # ========================================================

    if " - " in col_limpa:

        codigo_variavel, descricao_variavel = (
            col_limpa.split(
                " - ",
                1
            )
        )

        codigo_variavel = codigo_variavel.strip()
        descricao_variavel = descricao_variavel.strip()

    else:

        codigo_variavel = col_limpa
        descricao_variavel = col_limpa


    # ========================================================
    # 4.2 PROCURAR A DESCRIÇÃO NO DICIONÁRIO
    # ========================================================

    chave_busca = normalizar(
        descricao_variavel
    )


    cat_dict = dico_map.get(
        chave_busca,
        {}
    )


    # ========================================================
    # 4.3 CONTAR AS RESPOSTAS
    # ========================================================

    # NÃO usamos astype(str) aqui porque isso transforma NaN
    # em "nan" e dificulta identificar valores realmente nulos.

    counts = df_limpo[
        col
    ].value_counts(
        dropna=False
    )


    total = len(
        df_limpo
    )


    respostas_formatadas = []


    # ========================================================
    # 4.4 CONVERTER CÓDIGO DA RESPOSTA EM SIGNIFICADO
    # ========================================================

    for cod, qtd in counts.items():

        pct = (
            qtd / total
        ) * 100


        # ----------------------------------------------------
        # Tratamento de valores nulos
        # ----------------------------------------------------

        if pd.isna(cod):

            respostas_formatadas.append(
                f"[NaN] Valores Nulos / Em Branco (NaN): {pct:.2f}%"
            )

            continue


        # ----------------------------------------------------
        # Converte o código para string
        # ----------------------------------------------------

        cod_str = str(
            cod
        ).strip()


        # ----------------------------------------------------
        # Remove ".0" quando o código for algo como:
        #
        # 1.0 -> 1
        # 2.0 -> 2
        # ----------------------------------------------------

        if cod_str.endswith(".0"):

            cod_str = cod_str[:-2]


        # ----------------------------------------------------
        # Procura o significado no dicionário
        # ----------------------------------------------------

        rotulo = cat_dict.get(
            cod_str
        )


        # ----------------------------------------------------
        # Se não encontrou, tenta como número float
        # ----------------------------------------------------

        if rotulo is None:

            try:

                rotulo = cat_dict.get(
                    str(float(cod_str))
                )

            except (ValueError, TypeError):

                rotulo = None


        # ----------------------------------------------------
        # Se encontrou o significado
        # ----------------------------------------------------

        if rotulo is not None:

            respostas_formatadas.append(
                f"[{cod_str}] {rotulo}: {pct:.2f}%"
            )

        else:

            # Mantém o código mesmo quando não existe
            # significado no dicionário

            respostas_formatadas.append(
                f"[{cod_str}]: {pct:.2f}%"
            )


    # ========================================================
    # 4.5 ADICIONAR A VARIÁVEL À TABELA FINAL
    # ========================================================

    tabela_dados.append({

        "Pergunta / Variável":
            col_limpa,

        "Distribuição das Respostas":
            "\n".join(
                respostas_formatadas
            )
    })


# ============================================================
# 5. CRIAR O DATAFRAME DA TABELA
# ============================================================

df_tabela = pd.DataFrame(
    tabela_dados
)


# ============================================================
# 6. ESTILIZAR A TABELA
# ============================================================

styler = df_tabela.style.set_properties(
    **{
        "white-space": "pre-wrap",
        "text-align": "left"
    }
)


# ============================================================
# 7. EXIBIR O RESULTADO
# ============================================================

print(
    "========== EXIBIÇÃO DOS RESULTADOS NO IPYNB ==========\n"
)

display(
    HTML(
        styler.to_html()
    )
)
```
