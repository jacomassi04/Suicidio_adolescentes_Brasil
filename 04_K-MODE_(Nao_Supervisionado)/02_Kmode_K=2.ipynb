```python
import pandas as pd
import numpy as np
from kmodes.kmodes import KModes
import matplotlib.pyplot as plt
import seaborn as sns
import ipywidgets as widgets
from IPython.display import display, Markdown

# ============================================================
# 0. CARREGAR A BASE PENSE 2024
# ============================================================

from pathlib import Path
import pandas as pd

CAMINHO_BASE = Path(
    "/Users/claudiojacomassi/Documents/GitHub/"
    "Suicidio_adolescentes_Brasil/"
    "02_Tratamento_dados_database_PENSE_2024/"
    "02_Scripts_tratamento_dados_PENSE_2024/"
    "07_PENSE2024_IDADE_13_A_17_CATEGORY.feather"
)

if not CAMINHO_BASE.exists():
    raise FileNotFoundError(
        f"Arquivo não encontrado:\n{CAMINHO_BASE}"
    )

# Ler o arquivo Feather
df_base = pd.read_feather(CAMINHO_BASE)

print("Base carregada com sucesso!")
print(f"Dimensões: {df_base.shape[0]:,} linhas × {df_base.shape[1]:,} colunas")

display(df_base.head())

# ============================================================
# DEFINIR A MATRIZ PARA O K-MODES
# ============================================================

# Por padrão, utilizar todas as colunas.
# Se houver identificadores ou variáveis que não devam participar
# do clustering, selecione explicitamente as colunas abaixo.
X = df_base.copy()

print(f"\nDimensões de X: {X.shape}")
# ============================================================
# 1. PARÂMETROS DE ENTRADA DO K-MODES
# ============================================================
# Ajuste os controles abaixo e, em seguida, execute a próxima célula.

w_k = widgets.IntSlider(
    value=4, min=2, max=20, step=1,
    description="K (clusters):",
    continuous_update=False,
    style={"description_width": "initial"}
)

w_random_state = widgets.IntText(
    value=42,
    description="RANDOM_STATE:",
    style={"description_width": "initial"}
)

w_init = widgets.Dropdown(
    options=[
        ("Cao — baseado na densidade dos dados", "Cao"),
        ("Huang — baseado na frequência das categorias", "Huang"),
        ("Random — inicialização aleatória", "random")
    ],
    value="Huang",
    description="init:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="520px")
)

w_max_iter = widgets.IntSlider(
    value=100, min=10, max=500, step=10,
    description="max_iter:",
    continuous_update=False,
    style={"description_width": "initial"}
)

display(Markdown("### Defina os parâmetros do K-Modes"))
display(w_k, w_random_state, w_init, w_max_iter)
```

```python
# Execute esta célula depois de selecionar os parâmetros na célula anterior.
K = int(w_k.value)
RANDOM_STATE = int(w_random_state.value)
INIT = w_init.value
MAX_ITER = int(w_max_iter.value)

parametros_selecionados = pd.DataFrame({
    "Parâmetro": ["K (clusters)", "RANDOM_STATE", "init", "max_iter"],
    "Valor": [K, RANDOM_STATE, INIT, MAX_ITER]
})
display(Markdown("### Parâmetros selecionados"))
display(parametros_selecionados)

# ============================================================
# 2. PREPARAR OS DADOS
# ============================================================

# Verificar se a matriz de dados existe
if "X" in globals() and isinstance(X, pd.DataFrame):
    X = X.copy()

elif "df_base" in globals() and isinstance(df_base, pd.DataFrame):
    print("X não existe. Utilizando df_base como origem.")
    X = df_base.copy()

else:
    raise NameError(
        "Não foi encontrada a matriz X nem o DataFrame df_base. "
        "Execute primeiro a célula que carrega a base de dados "
        "e seleciona as variáveis para clustering."
    )

# Preservar a base original para exportação, quando disponível
if "df_base" in globals() and isinstance(df_base, pd.DataFrame):
    df_base_resultado = df_base.copy()
else:
    df_base_resultado = X.copy()

# Converter variáveis para strings
X = X.astype("string")

# Tratar valores ausentes
X = X.fillna("AUSENTE")

# Garantir compatibilidade com K-Modes
X = X.astype(str)

# Remover colunas constantes
colunas_constantes = [
    col for col in X.columns
    if X[col].nunique() <= 1
]

if colunas_constantes:
    print("Colunas constantes removidas:", colunas_constantes)
    X = X.drop(columns=colunas_constantes)

# Verificações
if X.empty or X.shape[1] == 0:
    raise ValueError(
        "Não há variáveis disponíveis para clustering."
    )

if len(X) <= K:
    raise ValueError(
        "O número de observações deve ser maior que K."
    )

print("Número de estudantes:", len(X))
print("Número de variáveis:", X.shape[1])
print("Número de clusters:", K)

# ============================================================
# 3. EXECUTAR O K-MODES
# ============================================================

modelo_kmodes = KModes(
    n_clusters=K,
    init=INIT,
    n_init=5,
    max_iter=MAX_ITER,
    random_state=RANDOM_STATE,
    verbose=1
)

clusters = modelo_kmodes.fit_predict(X)

# Adicionar os clusters à base de resultados
if df_base_resultado.index.equals(X.index):
    df_resultado = df_base_resultado.copy()
else:
    df_resultado = df_base_resultado.loc[X.index].copy()

df_resultado["cluster"] = clusters

# ============================================================
# 4. TAMANHO E DISTRIBUIÇÃO DOS CLUSTERS
# ============================================================

tamanhos = (
    df_resultado["cluster"]
    .value_counts()
    .sort_index()
)

resumo_clusters = pd.DataFrame({
    "cluster": tamanhos.index,
    "n_estudantes": tamanhos.values,
    "percentual": (
        tamanhos.values / len(df_resultado) * 100
    ).round(2)
})

print("\nRESUMO DOS CLUSTERS")
print(resumo_clusters.to_string(index=False))

# ============================================================
# 5. MODAS DE CADA CLUSTER
# ============================================================

modas_clusters = pd.DataFrame(
    modelo_kmodes.cluster_centroids_,
    columns=X.columns
)

modas_clusters.index.name = "cluster"

# ============================================================
# 5.1 SUBSTITUIR CÓDIGOS DAS MODAS PELOS RÓTULOS DA PLANILHA
# ============================================================
# A coluna "Nova codificacao" do Excel contém regras como:
#   3=Sim
#   1=Nao
#   0=Sem resposta
#
# O código identifica a variável pelo código no início do nome
# da coluna (ex.: "B01001A - Qual é o seu sexo?") e converte
# somente a tabela de modas. A base de estudantes e o modelo
# K-Modes permanecem com os códigos originais.

import re

CAMINHO_REGRAS_CODIFICACAO = Path(
    "/Users/claudiojacomassi/Documents/GitHub/"
    "Suicidio_adolescentes_Brasil/"
    "02_Tratamento_dados_database_PENSE_2024/"
    "02_Scripts_tratamento_dados_PENSE_2024/"
    "01_Colunas_recomendadas_pense_2024_final.xlsx"
)

if not CAMINHO_REGRAS_CODIFICACAO.exists():
    raise FileNotFoundError(
        "Não foi encontrado o Excel com os rótulos das categorias:\\n"
        f"{CAMINHO_REGRAS_CODIFICACAO}"
    )

regras_codificacao = pd.read_excel(
    CAMINHO_REGRAS_CODIFICACAO,
    header=1
)
regras_codificacao.columns = (
    regras_codificacao.columns.astype(str).str.strip()
)

coluna_codigo_excel = "Código da coluna"
coluna_rotulos_excel = "Nova codificacao"

for nome_coluna in [coluna_codigo_excel, coluna_rotulos_excel]:
    if nome_coluna not in regras_codificacao.columns:
        raise ValueError(
            f"A coluna '{nome_coluna}' não existe no Excel de codificação."
        )


def normalizar_codigo_categoria(valor):
    """Normaliza 1, 1.0 e '1' para a mesma chave textual."""
    if pd.isna(valor):
        return None

    texto = str(valor).strip()
    if not texto:
        return None

    try:
        numero = float(texto)
        if numero.is_integer():
            return str(int(numero))
    except (ValueError, TypeError):
        pass

    return texto


def criar_mapa_rotulos(texto_codificacao):
    """Converte texto como '3=Sim\\n1=Nao' em um dicionário."""
    mapa = {}

    if pd.isna(texto_codificacao):
        return mapa

    # Cada categoria normalmente ocupa uma linha; a expressão
    # também aceita espaços antes/depois do código e do rótulo.
    for linha in re.split(r"\n|;|,(?=\s*-?\d+\s*=)", str(texto_codificacao)):
        correspondencia = re.match(
            r"^\s*(-?\d+)\s*=\s*(.*?)\s*$",
            linha
        )
        if correspondencia:
            codigo = normalizar_codigo_categoria(correspondencia.group(1))
            rotulo = correspondencia.group(2).strip()
            if codigo is not None and rotulo:
                mapa[codigo] = rotulo

    return mapa


mapas_rotulos_por_variavel = {}

for _, linha_excel in regras_codificacao.iterrows():
    codigo_variavel = str(linha_excel[coluna_codigo_excel]).strip()
    if codigo_variavel and codigo_variavel.lower() != "nan":
        mapas_rotulos_por_variavel[codigo_variavel] = criar_mapa_rotulos(
            linha_excel[coluna_rotulos_excel]
        )

# Preservar uma cópia numérica para eventual conferência.
modas_clusters_codigos = modas_clusters.copy()
modas_clusters_rotuladas = modas_clusters.copy()

variaveis_sem_regra = []
codigos_sem_rotulo = {}

for nome_coluna in modas_clusters_rotuladas.columns:
    # As colunas da base têm o formato "CÓDIGO - descrição".
    codigo_variavel = str(nome_coluna).split(" - ", 1)[0].strip()
    mapa = mapas_rotulos_por_variavel.get(codigo_variavel, {})

    if not mapa:
        variaveis_sem_regra.append(codigo_variavel)
        continue

    valores_nao_mapeados = set()

    def substituir_por_rotulo(valor):
        codigo = normalizar_codigo_categoria(valor)

        if codigo is None:
            return valor

        if codigo in mapa:
            return mapa[codigo]

        valores_nao_mapeados.add(codigo)
        return valor  # não inventa rótulo quando o Excel não contém o código

    modas_clusters_rotuladas[nome_coluna] = (
        modas_clusters_rotuladas[nome_coluna]
        .map(substituir_por_rotulo)
    )

    if valores_nao_mapeados:
        codigos_sem_rotulo[codigo_variavel] = sorted(valores_nao_mapeados)

# Substituir a tabela principal de modas pela versão legível.
modas_clusters = modas_clusters_rotuladas

display(Markdown("### Modas dos clusters (rótulos da coluna 'Nova codificacao')"))
display(modas_clusters)

if variaveis_sem_regra:
    print(
        "\\nAviso: não foi encontrada regra 'Nova codificacao' para "
        f"{len(variaveis_sem_regra)} variável(is): "
        + ", ".join(variaveis_sem_regra)
    )

if codigos_sem_rotulo:
    print("\\nAviso: alguns códigos não tinham rótulo correspondente no Excel:")
    for codigo_variavel, codigos in codigos_sem_rotulo.items():
        print(f"  {codigo_variavel}: {codigos}")

# ============================================================
# 6. CUSTO FINAL DO MODELO
# ============================================================

print("\nCUSTO FINAL:", modelo_kmodes.cost_)
print("Número de iterações:", modelo_kmodes.n_iter_)

# ============================================================
# 7. GRÁFICO DO TAMANHO DOS CLUSTERS
# ============================================================

plt.figure(figsize=(9, 5))

sns.barplot(
    data=resumo_clusters,
    x="cluster",
    y="n_estudantes"
)

plt.title(f"K-Modes: tamanho dos clusters (K={K})")
plt.xlabel("Cluster")
plt.ylabel("Número de estudantes")
plt.tight_layout()
plt.show()

# ============================================================
# 8. EXIBIR RESULTADOS NO NOTEBOOK (SEM GERAR ARQUIVOS)
# ============================================================

display(Markdown(
    "### Resultados disponíveis no notebook\\n"
    "As tabelas abaixo são exibidas diretamente nesta saída. "
    "Nenhum arquivo CSV ou Feather é criado."
))

# Tabela-resumo (equivalente ao arquivo *_resumo.csv)
display(Markdown("#### Tabela-resumo dos clusters"))
display(resumo_clusters.style.format({"percentual": "{:.2f}%"}))

# Tabela de modas (equivalente ao arquivo *_modas.csv)
display(Markdown("#### Tabela de modas dos clusters"))
display(modas_clusters)
```

```python

# ============================================================
# 9. SILHOUETTE SCORE PARA K-MODES — PENSE2024
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import silhouette_score, silhouette_samples
from sklearn.metrics import pairwise_distances

# ------------------------------------------------------------
# 1. CONFIGURAÇÃO
# ------------------------------------------------------------

# K-Modes já deve ter sido treinado na célula anterior.
if "modelo_kmodes" not in globals():
    raise NameError("Execute primeiro a célula que treina o K-Modes.")

if "X" not in globals():
    raise NameError("A matriz X não foi encontrada.")

if "clusters" not in globals():
    raise NameError("Execute primeiro o modelo K-Modes.")

# Para bases grandes, calcular sobre uma amostra aleatória.
# 2.000 observações requerem aproximadamente 32 MB
# para a matriz de distâncias float64.
TAMANHO_AMOSTRA = min(2000, len(X))
RANDOM_STATE_SILHOUETTE = 42

if len(X) != len(clusters):
    raise ValueError("X e clusters precisam ter o mesmo número de linhas.")

# ------------------------------------------------------------
# 2. SELECIONAR AMOSTRA
# ------------------------------------------------------------

rng = np.random.default_rng(RANDOM_STATE_SILHOUETTE)

indices = rng.choice(
    len(X),
    size=TAMANHO_AMOSTRA,
    replace=False
)

X_sample = X.iloc[indices].astype(str).to_numpy()
labels_sample = np.asarray(clusters)[indices]

# O Silhouette exige pelo menos 2 clusters e menos clusters
# do que observações na amostra.
n_clusters_sample = np.unique(labels_sample).size

if n_clusters_sample < 2 or n_clusters_sample >= len(labels_sample):
    raise ValueError(
        "A amostra precisa conter pelo menos 2 clusters "
        "e menos clusters do que observações."
    )

print("Observações totais:", len(X))
print("Observações na amostra:", len(X_sample))
print("Variáveis:", X_sample.shape[1])
print("Clusters representados:", n_clusters_sample)

# ------------------------------------------------------------
# 3. MATRIZ DE DISTÂNCIA CATEGÓRICA
# ------------------------------------------------------------
# Distância de matching:
# 0 = todas as categorias são iguais
# 1 = todas as categorias são diferentes
#
# Cada variável recebe o mesmo peso.

D = pairwise_distances(
    X_sample,
    metric=lambda a, b: np.mean(a != b)
)

# ------------------------------------------------------------
# 4. CALCULAR SILHOUETTE SCORE
# ------------------------------------------------------------

silhouette_medio = silhouette_score(
    D,
    labels_sample,
    metric="precomputed"
)

silhouette_individual = silhouette_samples(
    D,
    labels_sample,
    metric="precomputed"
)

print("\n======================================")
print("SILHOUETTE SCORE — K-MODES")
print("======================================")
print(f"Score médio: {silhouette_medio:.4f}")

# ------------------------------------------------------------
# 5. SCORE POR CLUSTER
# ------------------------------------------------------------

resultado_silhouette = pd.DataFrame({
    "cluster": labels_sample,
    "silhouette": silhouette_individual
})

resumo_silhouette = (
    resultado_silhouette
    .groupby("cluster")["silhouette"]
    .agg(["mean", "median", "std", "count"])
    .reset_index()
)

resumo_silhouette.columns = [
    "cluster",
    "silhouette_medio",
    "silhouette_mediano",
    "desvio_padrao",
    "n_observacoes"
]

display(resumo_silhouette.round(4))

# ------------------------------------------------------------
# 6. GRÁFICO: SILHOUETTE POR CLUSTER
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=resultado_silhouette,
    x="cluster",
    y="silhouette"
)

plt.axhline(
    0,
    linestyle="--",
    linewidth=1
)

plt.title(
    f"Silhouette Score por Cluster — K-Modes (K={K})"
)
plt.xlabel("Cluster")
plt.ylabel("Silhouette Score")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 7. INTERPRETAÇÃO AUTOMÁTICA
# ------------------------------------------------------------

if silhouette_medio > 0.50:
    interpretacao = "Separação relativamente forte entre os clusters."
elif silhouette_medio > 0.25:
    interpretacao = "Separação moderada; examine os clusters individualmente."
elif silhouette_medio > 0:
    interpretacao = "Separação fraca; pode haver sobreposição entre clusters."
else:
    interpretacao = "Separação inadequada ou possível atribuição incorreta."

print("\nInterpretação inicial:")
print(interpretacao)

print(
    "\nATENÇÃO: o resultado utiliza uma amostra aleatória de "
    f"{TAMANHO_AMOSTRA} estudantes, e não necessariamente toda a base."
)
```

```python
# ============================================================
# CÉLULA 3 — GERAR UM PERFIL ALEATÓRIO E PREVER SEU CLUSTER
# Execute somente depois da Célula 2, que treina o K-Modes.
# ============================================================

import numpy as np
import pandas as pd
from IPython.display import display, Markdown

# ------------------------------------------------------------
# 1. CONFIGURAÇÃO
# ------------------------------------------------------------

# Sem semente fixa: um novo perfil aleatório é gerado a cada execução.
rng = np.random.default_rng()

# Número esperado de variáveis. O código verifica a quantidade
# efetivamente utilizada pelo modelo antes de gerar o perfil.
N_VARIAVEIS_ESPERADAS = 39

if "modelo_kmodes" not in globals():
    raise NameError(
        "O modelo K-Modes não foi encontrado. Execute primeiro a Célula 2."
    )

if "X" not in globals() or not isinstance(X, pd.DataFrame):
    raise NameError(
        "A matriz X não foi encontrada. Execute primeiro as células anteriores."
    )

# As colunas de X após o pré-processamento são as variáveis efetivamente
# utilizadas no treinamento (incluindo a eventual remoção de colunas constantes).
colunas_modelo = list(X.columns)

if len(colunas_modelo) != N_VARIAVEIS_ESPERADAS:
    print(
        f"AVISO: o modelo está usando {len(colunas_modelo)} variáveis, "
        f"e não {N_VARIAVEIS_ESPERADAS}. O perfil será gerado para "
        "todas as variáveis efetivamente usadas pelo modelo."
    )

# ------------------------------------------------------------
# 2. GERAR RESPOSTAS ALEATÓRIAS VÁLIDAS
# ------------------------------------------------------------
# Para cada variável, sorteia uma das categorias observadas em X.
# Isso evita criar códigos/categorias que não existam na base.
# O sorteio é uniforme entre categorias distintas, não proporcional
# à frequência de cada categoria na população.

respostas_perfil = {}

for coluna in colunas_modelo:
    categorias_validas = X[coluna].dropna().astype(str).unique().tolist()

    if not categorias_validas:
        raise ValueError(
            f"A variável '{coluna}' não possui categorias disponíveis."
        )

    respostas_perfil[coluna] = str(
        rng.choice(categorias_validas)
    )

# DataFrame com uma única observação, na mesma ordem do treinamento.
novo_perfil = pd.DataFrame(
    [[respostas_perfil[coluna] for coluna in colunas_modelo]],
    columns=colunas_modelo
).astype(str)

# ------------------------------------------------------------
# 3. PREVER O CLUSTER DO NOVO PERFIL
# ------------------------------------------------------------

cluster_previsto = int(modelo_kmodes.predict(novo_perfil)[0])

display(Markdown("## Perfil aleatório gerado"))
print("Perfil gerado com uma nova amostragem aleatória nesta execução.")
print(f"Número de variáveis usadas na previsão: {len(colunas_modelo)}")
display(novo_perfil.T.rename(columns={0: "Resposta sorteada"}))

display(Markdown("## Resultado da classificação"))
print(f"Cluster previsto: {cluster_previsto}")

# ------------------------------------------------------------
# 4. DISTÂNCIA A CADA CENTROIDE
# ------------------------------------------------------------
# No K-Modes, a dissimilaridade é baseada no número de categorias
# diferentes entre o perfil e cada centroide (distância de matching).
# Menor distância = maior proximidade ao centroide.
# A distância não é uma probabilidade de pertencer ao cluster.

centroides = np.asarray(modelo_kmodes.cluster_centroids_).astype(str)
perfil_array = novo_perfil.iloc[0].to_numpy(dtype=str)

distancias = np.sum(centroides != perfil_array, axis=1)

tabela_distancias = pd.DataFrame({
    "cluster": np.arange(len(distancias)),
    "dissimilaridade": distancias,
    "percentual_variaveis_diferentes": (
        distancias / len(colunas_modelo) * 100
    ).round(2)
}).sort_values("dissimilaridade", ascending=True).reset_index(drop=True)

display(Markdown("### Distância do perfil a cada centroide"))
display(tabela_distancias)

# ------------------------------------------------------------
# 5. INTERPRETAÇÃO SIMPLES
# ------------------------------------------------------------

distancia_escolhida = int(distancias[cluster_previsto])
percentual_diferente = distancia_escolhida / len(colunas_modelo) * 100

print(
    f"\nO perfil foi atribuído ao cluster {cluster_previsto}."
    f"\nEle difere do centroide desse cluster em "
    f"{distancia_escolhida} de {len(colunas_modelo)} variáveis "
    f"({percentual_diferente:.2f}%)."
)
print(
    "Observação: o perfil é sintético e sorteado aleatoriamente. "
    "A classificação não representa diagnóstico, risco individual "
    "ou probabilidade clínica."
)
```
