```python
# ============================================================
# K-MODES — DIAGNÓSTICO DE K E EXPORTAÇÃO DOS RESULTADOS
# Projeto: Suicidio_adolescentes_Brasil
# ============================================================

from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from kmodes.kmodes import KModes

# ------------------------------------------------------------
# 1. DEFINIR PASTA DE SAÍDA E ARQUIVO DE ENTRADA
# ------------------------------------------------------------

PASTA_SAIDA = Path(
    "/Users/claudiojacomassi/Documents/GitHub/"
    "Suicidio_adolescentes_Brasil/04_K-MODE_(Nao_Supervisionado)"
)

ARQUIVO_ENTRADA = Path(
    "/Users/claudiojacomassi/Documents/GitHub/"
    "Suicidio_adolescentes_Brasil/"
    "02_Tratamento_dados_database_PENSE_2024/"
    "02_Scripts_tratamento_dados_PENSE_2024/"
    "07_PENSE2024_IDADE_13_A_17_CATEGORY.feather"
)

PASTA_SAIDA.mkdir(parents=True, exist_ok=True)

if not ARQUIVO_ENTRADA.exists():
    raise FileNotFoundError(
        f"Arquivo não encontrado: {ARQUIVO_ENTRADA}"
    )

print("Pasta de saída:", PASTA_SAIDA)

# ------------------------------------------------------------
# 2. CARREGAR A BASE CATEGÓRICA (FEATHER)
# ------------------------------------------------------------

# Leitura alterada de read_parquet para read_feather
df_base = pd.read_feather(ARQUIVO_ENTRADA)

print("Base carregada com sucesso!")
print("Arquivo:", ARQUIVO_ENTRADA.name)
print(f"Registros: {len(df_base):,}")
print(f"Variáveis: {df_base.shape[1]:,}")

display(df_base.head())

# ------------------------------------------------------------
# 3. PREPARAR A MATRIZ PARA O K-MODES
# ------------------------------------------------------------

X = df_base.copy()

# Converter para strings e tratar valores ausentes
X = X.astype("string").fillna("AUSENTE")
X = X.astype(str)

# Remover colunas constantes
colunas_constantes = [
    coluna for coluna in X.columns
    if X[coluna].nunique(dropna=False) <= 1
]

if colunas_constantes:
    print(
        "Colunas constantes removidas:",
        colunas_constantes
    )
    X = X.drop(columns=colunas_constantes)

if X.shape[1] == 0:
    raise ValueError(
        "Nenhuma variável disponível para o K-Modes."
    )

print("\nMatriz preparada para o K-Modes:")
print("Dimensões:", X.shape)
print("Valores ausentes restantes:", X.isna().sum().sum())

print("\nPrimeiras variáveis:")
print(X.columns.tolist()[:20])

# ------------------------------------------------------------
# 4. DIAGNÓSTICO DE K
# ------------------------------------------------------------

K_MAX = 8
RANDOM_STATE = 42
N_INIT = 5
MAX_ITER = 100

resultados_k = []

for k in range(2, min(K_MAX, len(X) - 1) + 1):

    print(f"Executando K-Modes para K={k}...")

    modelo = KModes(
        n_clusters=k,
        init="Huang",
        n_init=N_INIT,
        max_iter=MAX_ITER,
        random_state=RANDOM_STATE,
        verbose=0
    )

    labels = modelo.fit_predict(X)

    tamanhos = pd.Series(labels).value_counts()

    resultados_k.append({
        "K": k,
        "Custo": modelo.cost_,
        "Menor_cluster": int(tamanhos.min()),
        "Maior_cluster": int(tamanhos.max()),
        "Menor_cluster_%": round(
            tamanhos.min() / len(X) * 100, 2
        ),
        "Clusters_vazios": k - len(tamanhos)
    })

diagnostico_k = pd.DataFrame(resultados_k)

display(diagnostico_k)

# ------------------------------------------------------------
# 5. GRÁFICO DO MÉTODO DO COTOVELO
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

sns.lineplot(
    data=diagnostico_k,
    x="K",
    y="Custo",
    marker="o"
)

plt.title("K-Modes — Custo por número de clusters")
plt.xlabel("Número de clusters (K)")
plt.ylabel("Custo total")
plt.xticks(diagnostico_k["K"])
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig(
    PASTA_SAIDA / "grafico_cotovelo_KModes.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ------------------------------------------------------------
# 6. SALVAR DIAGNÓSTICO E MATRIZ UTILIZADA
# ------------------------------------------------------------

diagnostico_k.to_csv(
    PASTA_SAIDA / "diagnostico_K.csv",
    index=False,
    encoding="utf-8-sig"
)

# Exportação alterada para formato Feather
X.to_feather(PASTA_SAIDA / "matriz_categorica_KModes.feather")

print("\nDiagnóstico concluído.")
print("Arquivos salvos em:")
print(PASTA_SAIDA)

print("\nArquivos gerados:")
print("- diagnostico_K.csv")
print("- grafico_cotovelo_KModes.png")
print("- matriz_categorica_KModes.feather")
```

```python

# ============================================================
# K-MODES — DIAGNÓSTICO DE K (SILHOUETTE SCORE)
# Projeto: Suicidio_adolescentes_Brasil
# Exibe o gráfico e a tabela diretamente no Jupyter Notebook
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path
from kmodes.kmodes import KModes
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import OrdinalEncoder

# ------------------------------------------------------------
# 1. DEFINIR ARQUIVO DE ENTRADA
# ------------------------------------------------------------

ARQUIVO_ENTRADA = Path(
    "/Users/claudiojacomassi/Documents/GitHub/"
    "Suicidio_adolescentes_Brasil/"
    "02_Tratamento_dados_database_PENSE_2024/"
    "02_Scripts_tratamento_dados_PENSE_2024/"
    "07_PENSE2024_IDADE_13_A_17_CATEGORY.feather"
)

if not ARQUIVO_ENTRADA.exists():
    raise FileNotFoundError(
        f"Arquivo não encontrado: {ARQUIVO_ENTRADA}"
    )

# ------------------------------------------------------------
# 2. CARREGAR A BASE CATEGÓRICA (FEATHER)
# ------------------------------------------------------------

df_base = pd.read_feather(ARQUIVO_ENTRADA)

print("Base carregada com sucesso!")
print(f"Registros: {len(df_base):,}")
print(f"Variáveis: {df_base.shape[1]:,}")

# ------------------------------------------------------------
# 3. PREPARAR A MATRIZ PARA O K-MODES E SCORING
# ------------------------------------------------------------

X = df_base.copy()
X = X.astype("string").fillna("AUSENTE")
X = X.astype(str)

# Remover variáveis constantes
colunas_constantes = [
    coluna for coluna in X.columns
    if X[coluna].nunique(dropna=False) <= 1
]

if colunas_constantes:
    print(
        f"\nVariáveis constantes removidas: "
        f"{len(colunas_constantes)}"
    )
    X = X.drop(columns=colunas_constantes)

if X.shape[1] == 0:
    raise ValueError(
        "Não restaram variáveis após a remoção das "
        "colunas constantes."
    )

# Codificação utilizada apenas para o Silhouette Score
X_encoded = OrdinalEncoder().fit_transform(X)

print("\nMatriz preparada para o K-Modes.")
print(f"Registros utilizados: {len(X):,}")
print(f"Variáveis utilizadas: {X.shape[1]:,}")

# ------------------------------------------------------------
# 4. DIAGNÓSTICO DE K COM SILHOUETTE SCORE
# ------------------------------------------------------------

K_MAX = 8
RANDOM_STATE = 42
N_INIT = 5
MAX_ITER = 100
SAMPLE_SIZE = 15000

resultados_sil = []

print(
    "\nIniciando cálculo do Silhouette Score "
    f"(amostra de até {SAMPLE_SIZE:,} registros)...\n"
)

for k in range(2, min(K_MAX, len(X) - 1) + 1):

    print(f"Ajustando K-Modes para K={k}...")

    modelo = KModes(
        n_clusters=k,
        init="Huang",
        n_init=N_INIT,
        max_iter=MAX_ITER,
        random_state=RANDOM_STATE,
        verbose=0
    )

    # Treinar o modelo com os dados categóricos originais
    labels = modelo.fit_predict(X)

    # Calcular o Silhouette Score com distância de Hamming
    sil_score = silhouette_score(
        X_encoded,
        labels,
        metric="hamming",
        sample_size=min(SAMPLE_SIZE, len(X)),
        random_state=RANDOM_STATE
    )

    resultados_sil.append({
        "K": k,
        "Silhouette_Score": sil_score
    })

    print(
        f"--> Silhouette Score (K={k}): "
        f"{sil_score:.4f}\n"
    )

# Criar e exibir a tabela no notebook
df_sil = pd.DataFrame(resultados_sil)

display(df_sil)

# ------------------------------------------------------------
# 5. EXIBIR O GRÁFICO NO JUPYTER NOTEBOOK
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

sns.lineplot(
    data=df_sil,
    x="K",
    y="Silhouette_Score",
    marker="s",
    color="darkorange",
    linewidth=2,
    markersize=8
)

plt.title(
    "K-Modes — Silhouette Score por número de clusters",
    fontsize=14
)

plt.xlabel("Número de clusters (K)", fontsize=12)
plt.ylabel("Silhouette Score (Hamming)", fontsize=12)

plt.xticks(df_sil["K"])
plt.grid(True, alpha=0.3)
plt.tight_layout()

# Exibir somente na saída da célula
plt.show()

# ------------------------------------------------------------
# 6. CONCLUSÃO
# ------------------------------------------------------------

if not df_sil.empty:
    melhor_linha = df_sil.loc[
        df_sil["Silhouette_Score"].idxmax()
    ]

    print("\nDiagnóstico Silhouette concluído.")
    print(
        f"Maior Silhouette Score: "
        f"K={int(melhor_linha['K'])}, "
        f"Score={melhor_linha['Silhouette_Score']:.4f}"
    )
else:
    print("Não foi possível calcular os resultados.")
```
