Importando dados da base SIM (DataSus) 2024 - Dados sobre mortalidade em geral

```python
df_SIMSUS = pd.read_parquet('MORTALIDADE_geral_2024.parquet', engine='pyarrow')

print(df_SIMSUS.head())

df_SIMSUS.shape

df_SIMSUS.info()
```

Copiar apenas as Colunas desejadas dos DataFrames Pense e SimSUS

```python
colunas_SimSus = ['CODMUNRES', 'IDADE', 'SEXO', 'RACACOR', 'CAUSABAS', 'CIRCOBITO', 'LINHAA', 'LINHAB', 'LINHAC', 'LINHAD', 'LINHAII', 'LOCOCOR', 'ESC2010', 'SERIESCFAL', 'ESCMAE2010']

df_SimSus_colunas = df_SIMSUS[colunas_SimSus].copy()

print(df_SimSus_colunas.head())

df_SimSus_colunas.shape

df_SimSus_colunas.info()
```
