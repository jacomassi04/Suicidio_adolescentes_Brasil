```python
### Regressão linear múltipla — PENSE 2024

# ============================================================
# 1. IMPORTAR BIBLIOTECAS
# ============================================================

from pathlib import Path
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score
)

import statsmodels.api as sm

warnings.filterwarnings("ignore", category=FutureWarning)

# ============================================================
# 2. CAMINHOS
# ============================================================

ARQUIVO = Path(
    "/Users/claudiojacomassi/Documents/GitHub/"
    "Suicidio_adolescentes_Brasil/"
    "02_Tratamento_dados_database_PENSE_2024/"
    "02_Scripts_tratamento_dados_PENSE_2024/"
    "07_PENSE2024_IDADE_13_A_17_CATEGORY.parquet"
)

PASTA_SAIDA = Path(
    "/Users/claudiojacomassi/Documents/GitHub/"
    "Suicidio_adolescentes_Brasil/"
    "05_Regressao_logistica_(Supervisionado)"
)

PASTA_SAIDA.mkdir(parents=True, exist_ok=True)

# ============================================================
# 3. CARREGAR A BASE
# ============================================================

if not ARQUIVO.exists():
    raise FileNotFoundError(
        f"Arquivo não encontrado: {ARQUIVO}"
    )

df = pd.read_parquet(ARQUIVO)

print("=" * 70)
print("BASE DE DADOS PENSE 2024")
print("=" * 70)
print(f"Linhas: {df.shape[0]:,}")
print(f"Colunas: {df.shape[1]:,}")
print(f"Memória: {df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")

print("\nTipos de dados:")
print(df.dtypes.value_counts())

print("\nPrimeiras linhas:")
display(df.head())

# ============================================================
# 4. IDENTIFICAR AS VARIÁVEIS
# ============================================================

print("\nNomes das colunas disponíveis:\n")

for i, coluna in enumerate(df.columns, start=1):
    print(f"{i:03d} - {coluna}")

# ------------------------------------------------------------
# IMPORTANTE:
# Substitua pelo nome exato da variável dependente.
# Exemplo hipotético: TARGET = "NOME_DA_VARIAVEL"
# ------------------------------------------------------------

TARGET = "B12009 - NOS ÚLTIMOS 12 MESES, alguma vez você sentiu vontade de se machucar de propósito?"

if TARGET not in df.columns:
    print("\nDefina TARGET com uma coluna existente.")
    print("O programa será interrompido antes do ajuste.")
    raise ValueError(
        "Altere TARGET para o nome real da variável dependente."
    )

# ============================================================
# 5. PREPARAR A BASE
# ============================================================

dados = df.copy()

# Remover linhas sem valor observado no desfecho.
dados = dados.loc[dados[TARGET].notna()].copy()

# A regressão linear requer uma variável dependente numérica.
# Se TARGET for categórica, verifique sua codificação antes
# de prosseguir. Não converta rótulos arbitrários em números.

if not pd.api.types.is_numeric_dtype(dados[TARGET]):
    raise TypeError(
        f"A variável {TARGET} não é numérica. "
        "Verifique o desfecho e escolha o tipo de modelo adequado."
    )

# Excluir preditores constantes e a própria variável dependente.
preditores = [
    col for col in dados.columns
    if col != TARGET and dados[col].nunique(dropna=True) > 1
]

if not preditores:
    raise ValueError("Não há variáveis preditoras disponíveis.")

X = dados[preditores].copy()
y = dados[TARGET].astype(float)

# Remover preditores totalmente ausentes.
colunas_validas = X.columns[X.notna().any()].tolist()
X = X[colunas_validas]

# Identificar variáveis numéricas e categóricas.
# Booleanos são tratados como categóricos.
num_cols = X.select_dtypes(include=["number"]).columns.tolist()

cat_cols = [
    col for col in X.columns
    if col not in num_cols
]

print("\nVariável dependente:", TARGET)
print("Número de preditores:", X.shape[1])
print("Preditores numéricos:", len(num_cols))
print("Preditores categóricos:", len(cat_cols))

# ============================================================
# 6. DIVIDIR TREINO E TESTE
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# ============================================================
# 7. TRATAR VARIÁVEIS E CODIFICAR CATEGORIAS
# ============================================================

transformadores = []

if num_cols:
    pipeline_num = Pipeline([
        ("imputacao", SimpleImputer(strategy="median"))
    ])

    transformadores.append(
        ("numericas", pipeline_num, num_cols)
    )

if cat_cols:
    pipeline_cat = Pipeline([
        ("imputacao", SimpleImputer(strategy="most_frequent")),
        ("codificacao", OneHotEncoder(
            handle_unknown="ignore",
            drop="first"
        ))
    ])

    transformadores.append(
        ("categoricas", pipeline_cat, cat_cols)
    )

preprocessador = ColumnTransformer(
    transformers=transformadores,
    remainder="drop"
)

# ============================================================
# 8. AJUSTAR O MODELO
# ============================================================

modelo = Pipeline([
    ("preprocessamento", preprocessador),
    ("regressao", LinearRegression())
])

modelo.fit(X_train, y_train)

# ============================================================
# 9. PREDIÇÕES
# ============================================================

y_pred = modelo.predict(X_test)

# ============================================================
# 10. AVALIAR O MODELO
# ============================================================

rmse = np.sqrt(mean_squared_error(y_test, y_pred))
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

metricas = pd.DataFrame({
    "Metrica": ["R2", "RMSE", "MAE"],
    "Valor": [r2, rmse, mae]
})

print("\n" + "=" * 70)
print("AVALIAÇÃO DO MODELO")
print("=" * 70)
display(metricas)

metricas.to_csv(
    PASTA_SAIDA / "metricas_modelo.csv",
    index=False
)

# ============================================================
# 11. COEFICIENTES DO MODELO
# ============================================================

nomes_features = (
    modelo.named_steps["preprocessamento"]
    .get_feature_names_out()
)

coeficientes = pd.DataFrame({
    "Variavel": nomes_features,
    "Coeficiente": modelo.named_steps["regressao"].coef_
})

coeficientes["Valor_absoluto"] = (
    coeficientes["Coeficiente"].abs()
)

coeficientes = coeficientes.sort_values(
    "Valor_absoluto",
    ascending=False
)

print("\nCoeficientes da regressão:")
display(coeficientes.head(30))

coeficientes.to_csv(
    PASTA_SAIDA / "coeficientes.csv",
    index=False
)

# ============================================================
# 12. GRÁFICO DOS COEFICIENTES
# ============================================================

top_coef = coeficientes.head(20).sort_values("Coeficiente")

plt.figure(figsize=(12, 8))

sns.barplot(
    data=top_coef,
    x="Coeficiente",
    y="Variavel"
)

plt.axvline(0, color="black", linewidth=1)
plt.title("20 maiores coeficientes em valor absoluto")
plt.xlabel("Coeficiente estimado")
plt.ylabel("Variável")
plt.tight_layout()

plt.savefig(
    PASTA_SAIDA / "grafico_coeficientes.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 13. VALORES OBSERVADOS VS. PREDITOS
# ============================================================

plt.figure(figsize=(8, 8))

sns.scatterplot(
    x=y_test,
    y=y_pred,
    alpha=0.5
)

limite_min = min(y_test.min(), y_pred.min())
limite_max = max(y_test.max(), y_pred.max())

plt.plot(
    [limite_min, limite_max],
    [limite_min, limite_max],
    linestyle="--",
    color="red"
)

plt.title("Valores observados vs. preditos")
plt.xlabel("Valor observado")
plt.ylabel("Valor predito")
plt.tight_layout()

plt.savefig(
    PASTA_SAIDA / "observado_vs_predito.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 14. DISTRIBUIÇÃO DOS RESÍDUOS
# ============================================================

residuos = y_test.to_numpy() - y_pred

plt.figure(figsize=(10, 6))

sns.histplot(
    residuos,
    kde=True
)

plt.axvline(0, linestyle="--", color="red")
plt.title("Distribuição dos resíduos")
plt.xlabel("Resíduo")
plt.ylabel("Frequência")
plt.tight_layout()

plt.savefig(
    PASTA_SAIDA / "distribuicao_residuos.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 15. RESÍDUOS VS. VALORES PREDITOS
# ============================================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    x=y_pred,
    y=residuos,
    alpha=0.5
)

plt.axhline(0, linestyle="--", color="red")
plt.title("Resíduos vs. valores preditos")
plt.xlabel("Valor predito")
plt.ylabel("Resíduo")
plt.tight_layout()

plt.savefig(
    PASTA_SAIDA / "residuos_vs_preditos.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 16. EXPORTAR PREVISÕES
# ============================================================

previsoes = pd.DataFrame({
    "Valor_observado": y_test.to_numpy(),
    "Valor_predito": y_pred,
    "Residuo": residuos
}, index=y_test.index)

previsoes.to_csv(
    PASTA_SAIDA / "previsoes_teste.csv",
    index_label="Indice_original"
)

# ============================================================
# 17. RELATÓRIO FINAL
# ============================================================

print("\n" + "=" * 70)
print("ANÁLISE CONCLUÍDA")
print("=" * 70)

print(f"Variável dependente: {TARGET}")
print(f"Amostra de treino: {len(X_train):,}")
print(f"Amostra de teste: {len(X_test):,}")
print(f"R²: {r2:.4f}")
print(f"RMSE: {rmse:.4f}")
print(f"MAE: {mae:.4f}")
print(f"\nResultados salvos em:\n{PASTA_SAIDA}")


```
