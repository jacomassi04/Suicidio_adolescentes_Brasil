```python

# ============================================================
# PENSE 2024 - MAPAS DE CALOR DE ASSOCIACAO CATEGORICA
# Base: adolescentes de 13 a 17 anos
# Entrada: Feather
# Metodo: V de Cramer com correcao de vies
# ============================================================

# ============================================================
# 1. IMPORTAR BIBLIOTECAS
# ============================================================

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from scipy.stats import chi2_contingency
from itertools import combinations

# ============================================================
# 2. DEFINIR CAMINHO DA BASE DE ENTRADA
# ============================================================

# Ajuste este caminho para o local onde o arquivo Feather está salvo.
caminho_base = "/Users/claudiojacomassi/Documents/GitHub/Suicidio_adolescentes_Brasil/02_Tratamento_dados_database_PENSE_2024/02_Scripts_tratamento_dados_PENSE_2024/07_PENSE2024_IDADE_13_A_17_CATEGORY.feather"

if not __import__("os").path.exists(caminho_base):
    raise FileNotFoundError(
        f"Arquivo Feather nao encontrado:\n{caminho_base}\n"
        "Atualize a variavel caminho_base para o caminho correto."
    )

print("Arquivo de entrada:", caminho_base)
print("Modo de execução: resultados somente no notebook; nenhum arquivo de saída será gravado.")

# ============================================================
# 3. CARREGAR A BASE FEATHER
# ============================================================

df = pd.read_feather(caminho_base)

print("\n" + "=" * 65)
print("DIAGNOSTICO DA BASE")
print("=" * 65)

print(f"Linhas: {df.shape[0]:,}")
print(f"Colunas: {df.shape[1]:,}")
print("\nTipos originais:")
print(df.dtypes.value_counts())

# ============================================================
# 4. PREPARAR AS VARIAVEIS CATEGORICAS
# ============================================================

# A base CATEGORY e considerada categorica.
# Exclua identificadores, pesos amostrais e variaveis
# que nao sejam categorias, conforme o dicionario da PENSE.

colunas_excluir = [
    # Exemplo:
    # "ID_ALUNO",
    # "PESO_AMOSTRAL"
]

colunas_categoricas = [
    coluna for coluna in df.columns
    if coluna not in colunas_excluir
]

if len(colunas_categoricas) < 2:
    raise ValueError(
        "Sao necessarias pelo menos duas variaveis."
    )

df_cat = df[colunas_categoricas].copy()

for coluna in colunas_categoricas:
    df_cat[coluna] = df_cat[coluna].astype("category")

print(
    f"\nVariaveis selecionadas: {len(colunas_categoricas)}"
)

# Mostrar diagnostico das variaveis no notebook
from IPython.display import display

diagnostico = pd.DataFrame({
    "Variavel": colunas_categoricas,
    "Tipo_original": [str(df[col].dtype) for col in colunas_categoricas],
    "Categorias_observadas": [df_cat[col].nunique(dropna=True) for col in colunas_categoricas],
    "Valores_ausentes": [df_cat[col].isna().sum() for col in colunas_categoricas]
})

print("\nDiagnóstico das variáveis:")
display(diagnostico)

# ============================================================
# 5. FUNCAO V DE CRAMER COM CORRECAO DE VIES
# ============================================================

def cramers_v(x, y):
    """
    Calcula V de Cramer corrigido para vies de Bergsma.
    Observacoes ausentes sao excluidas por par.
    """

    dados = pd.DataFrame({
        "x": x,
        "y": y
    }).dropna()

    if len(dados) < 2:
        return np.nan

    tabela = pd.crosstab(
        dados["x"],
        dados["y"]
    )

    r, k = tabela.shape
    n = tabela.to_numpy().sum()

    if r < 2 or k < 2 or n <= 1:
        return np.nan

    qui2 = chi2_contingency(
        tabela,
        correction=False
    )[0]

    phi2 = qui2 / n

    phi2_corr = max(
        0,
        phi2 - ((k - 1) * (r - 1)) / (n - 1)
    )

    r_corr = r - ((r - 1) ** 2) / (n - 1)
    k_corr = k - ((k - 1) ** 2) / (n - 1)

    denominador = min(
        k_corr - 1,
        r_corr - 1
    )

    if denominador <= 0:
        return np.nan

    return float(
        np.clip(
            np.sqrt(phi2_corr / denominador),
            0,
            1
        )
    )

# ============================================================
# 6. CALCULAR A MATRIZ COMPLETA DE ASSOCIACAO
# ============================================================

n_variaveis = len(colunas_categoricas)

matriz = pd.DataFrame(
    np.nan,
    index=colunas_categoricas,
    columns=colunas_categoricas
)

total_pares = n_variaveis * (n_variaveis - 1) // 2

print("\n" + "=" * 65)
print("CALCULANDO V DE CRAMER")
print("=" * 65)

for contador, (i, j) in enumerate(
    combinations(range(n_variaveis), 2),
    start=1
):

    col1 = colunas_categoricas[i]
    col2 = colunas_categoricas[j]

    valor = cramers_v(
        df_cat[col1],
        df_cat[col2]
    )

    matriz.loc[col1, col2] = valor
    matriz.loc[col2, col1] = valor

    if contador % 100 == 0 or contador == total_pares:
        print(
            f"Pares calculados: "
            f"{contador:,}/{total_pares:,}"
        )

# Diagonal principal
for coluna in colunas_categoricas:
    if df_cat[coluna].nunique(dropna=True) > 1:
        matriz.loc[coluna, coluna] = 1.0

print("\nMatriz completa de V de Cramer calculada:")
display(matriz)

# ============================================================
# 7. PRIMEIRO GRAFICO: MAPA DE CALOR COMPLETO
# ============================================================

sns.set_theme(style="white")

largura = max(14, min(40, n_variaveis * 0.45))
altura = max(12, min(36, n_variaveis * 0.40))

fig, ax = plt.subplots(
    figsize=(largura, altura)
)

sns.heatmap(
    matriz,
    ax=ax,
    cmap="RdYlBu_r",
    vmin=0,
    vmax=1,
    square=True,
    linewidths=0.1,
    mask=matriz.isna(),
    cbar_kws={
        "label": "V de Cramer corrigido",
        "shrink": 0.75
    }
)

ax.set_title(
    "PENSE 2024 - Associacao entre variaveis categoricas\n"
    "V de Cramer com correcao de vies",
    fontsize=15,
    pad=20
)

ax.set_xlabel("Variaveis")
ax.set_ylabel("Variaveis")

plt.xticks(rotation=90, fontsize=7)
plt.yticks(rotation=0, fontsize=7)
plt.tight_layout()

plt.show()
plt.close(fig)

# ============================================================
# 8. RANKING DOS PARES DE VARIAVEIS
# ============================================================

pares = []

for i, j in combinations(range(n_variaveis), 2):

    col1 = colunas_categoricas[i]
    col2 = colunas_categoricas[j]

    valor = matriz.loc[col1, col2]

    if pd.notna(valor):
        pares.append({
            "Variavel_1": col1,
            "Variavel_2": col2,
            "V_Cramer": valor
        })

df_pares = pd.DataFrame(
    pares,
    columns=[
        "Variavel_1",
        "Variavel_2",
        "V_Cramer"
    ]
)

df_pares = df_pares.sort_values(
    "V_Cramer",
    ascending=False
).reset_index(drop=True)

print("\nRanking completo das associações (primeiras 50 linhas):")
display(df_pares.head(50))

# ============================================================
# 9. GRAFICO E TABELA DAS MAIORES ASSOCIACOES
# ============================================================

# Escolha quantas associacoes deseja visualizar:
numero_associacoes = 20  # Altere para 5, 20, 30, etc.

# Selecionar as N maiores associacoes do ranking
numero_associacoes = max(1, int(numero_associacoes))
top_associacoes = df_pares.head(numero_associacoes).copy()

print("\n" + "=" * 65)
print(f"TOP {len(top_associacoes)} MAIORES ASSOCIACOES")
print("=" * 65)

if not top_associacoes.empty:

    # Identificar as variaveis presentes nos pares selecionados
    variaveis_top = list(dict.fromkeys(
        top_associacoes["Variavel_1"].tolist()
        + top_associacoes["Variavel_2"].tolist()
    ))

    # Matriz reduzida para o mapa de calor
    matriz_top = matriz.loc[
        variaveis_top,
        variaveis_top
    ].copy()

    # Criar mascara: True significa ocultar a celula
    mascara = pd.DataFrame(
        True,
        index=matriz_top.index,
        columns=matriz_top.columns,
        dtype=bool
    )

    # Mostrar somente as celulas dos pares selecionados
    for _, linha in top_associacoes.iterrows():
        v1 = linha["Variavel_1"]
        v2 = linha["Variavel_2"]
        mascara.loc[v1, v2] = False
        mascara.loc[v2, v1] = False

    # Manter a diagonal principal visivel
    for i in range(min(mascara.shape)):
        mascara.iloc[i, i] = False

    # Ocultar valores ausentes
    mascara = mascara | matriz_top.isna()

    # Criar e mostrar o grafico
    fig, ax = plt.subplots(
        figsize=(
            max(10, len(variaveis_top) * 0.75),
            max(8, len(variaveis_top) * 0.65)
        )
    )

    sns.heatmap(
        matriz_top,
        ax=ax,
        cmap="RdYlBu_r",
        vmin=0,
        vmax=1,
        annot=True,
        fmt=".2f",
        square=True,
        linewidths=0.4,
        mask=mascara,
        cbar_kws={"label": "V de Cramer corrigido"}
    )

    ax.set_title(
        f"PENSE 2024 - Top {len(top_associacoes)} maiores associacoes\n"
        "V de Cramer com correcao de vies",
        fontsize=14,
        pad=18
    )
    ax.set_xlabel("Variaveis")
    ax.set_ylabel("Variaveis")
    plt.xticks(rotation=90, fontsize=8)
    plt.yticks(rotation=0, fontsize=8)
    plt.tight_layout()
    plt.show()
    plt.close(fig)

    # A tabela e exibida DEPOIS do grafico
    print(f"\nTabela: Top {len(top_associacoes)} associacoes")
    tabela_top = top_associacoes.copy()
    tabela_top.insert(0, "Posicao", range(1, len(tabela_top) + 1))
    display(
        tabela_top.style.format({"V_Cramer": "{:.4f}"})
    )

else:
    matriz_top = pd.DataFrame()
    print("Nenhum par valido encontrado para gerar o grafico e a tabela.")

# 10. RESUMO FINAL
# ============================================================

print("\n" + "=" * 65)
print("ANALISE CONCLUIDA")
print("=" * 65)

print(f"Numero de linhas: {len(df_cat):,}")
print(f"Variaveis analisadas: {n_variaveis:,}")
print(f"Pares possiveis: {total_pares:,}")
print(f"Pares validos: {len(df_pares):,}")
print(f"Pares exibidos no grafico e na tabela: {len(top_associacoes):,}")

print("\nResultados mantidos na memória do notebook:")
print(" - diagnostico: diagnóstico das variáveis")
print(" - matriz: matriz completa de V de Cramer")
print(" - df_pares: ranking completo das associações")
print(" - top_associacoes: associacoes selecionadas pelo parametro numero_associacoes")
print(" - matriz_top: matriz reduzida usada no mapa de calor das associacoes selecionadas")
print("\nNenhum arquivo de resultado foi gravado em disco.")
```
