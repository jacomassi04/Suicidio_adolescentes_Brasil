```python

import pandas as pd
from pathlib import Path

# Caminhos dos arquivos
arquivo_entrada = Path("01_PENSE2024_MICRODADOS.parquet")
arquivo_saida = Path("02_PENSE2024_MICRODADOS_SELECIONADAS.parquet")

# Lista de variáveis extraídas do documento (30 originais)
# variaveis_desejadas = [
#     "B12009",
#     "B12008",
#     "B12005",
#     "B12006",
#     "B12004",
#     "B12007",
#     "B12003",
#     "B07004",
#     "B07002",
#     "B07007B",
#     "B07011A",
#     "B07013",
#     "B09003A",
#     "B09016A1",
#     "B09016A2",
#     "B05003",
#     "B06007",
#     "B06008",
#     "B04003",
#     "B09001",
#     "B09002",
#     "B01001A",
#     "IDADE_AGREG",
#     "B01002",
#     "REGIAO",
#     "UF",
#     "CAPITAL",
#     "B01008B",
#     "B01015B",
#     "B01016",
#     "B03010C",
#     "B03006B"
# ]

# Lista de variáveis extraídas do documento (novas variaveis apos limitar das 97 variaveis apeans s 35 que possuem mennos que 5% de dados ou sem resposta ou abandono ou nulas)
variaveis_desejadas = [
    "B12009",
    "B12008",
    "B12004",
    "B12006",
    "B12007",
    "B12003",
    "B07006",
    "B07007B",
    "B07011A",
    "B07012A",
    "B07013",
    "B07009",
    "B07016",
    "B09003A",
    "B09010A1",
    "B09001",
    "B09002",
    "B07002",
    "B07004",
    "B07005",
    "B01006",
    "B01007",
    "B01001A",
    "IDADE_AGREG",
    "B01002",
    "REGIAO",
    "UF",
    "CAPITAL",
    "B01015B",
    "B01016",
    "B03010C",
    "B03006B",
    "E01P117",
    "E01P8505",
    "E01P8510"
]

# Lê o arquivo Parquet
df = pd.read_parquet(arquivo_entrada)

# Verifica quais variáveis existem na base
variaveis_encontradas = [
    coluna for coluna in variaveis_desejadas
    if coluna in df.columns
]

# Identifica variáveis não encontradas
variaveis_ausentes = [
    coluna for coluna in variaveis_desejadas
    if coluna not in df.columns
]

# Cria a nova base somente com as variáveis selecionadas
df_reduzido = df[variaveis_encontradas].copy()

# Grava o novo arquivo Parquet
df_reduzido.to_parquet(
    arquivo_saida,
    engine="pyarrow",
    compression="snappy",
    index=False
)

# Relatório
print("Processamento concluído!")
print("-" * 50)

print(f"Linhas originais: {df.shape[0]:,}")
print(f"Colunas originais: {df.shape[1]}")
print(f"Colunas selecionadas: {df_reduzido.shape[1]}")
print(f"Linhas preservadas: {df_reduzido.shape[0]:,}")

print("\nVariáveis selecionadas:")
print(variaveis_encontradas)

print("\nVariáveis não encontradas:")
print(variaveis_ausentes)

print(f"\nArquivo gerado: {arquivo_saida.resolve()}")
print(f"Tamanho do arquivo: {arquivo_saida.stat().st_size / (1024**2):.2f} MB")
```
