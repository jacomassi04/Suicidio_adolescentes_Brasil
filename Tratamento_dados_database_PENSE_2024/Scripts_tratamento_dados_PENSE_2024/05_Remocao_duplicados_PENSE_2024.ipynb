```python
import pandas as pd
from pathlib import Path

# Caminho do arquivo original
arquivo = Path("04_PENSE2024_SEM_NAN.parquet")

# Carrega o DataFrame
df = pd.read_parquet(arquivo)

# Verifica duplicatas considerando todas as colunas
total_linhas = len(df)
duplicadas = df.duplicated(keep="first").sum()
linhas_envolvidas = df.duplicated(keep=False).sum()

# Calcula o percentual de duplicatas
percentual = (duplicadas / total_linhas) * 100 if total_linhas else 0

# Exibe o resultado
print(f"Total de linhas: {total_linhas:,}")
print(f"Linhas duplicadas excedentes: {duplicadas:,}")
print(f"Total de linhas envolvidas em duplicação: {linhas_envolvidas:,}")
print(f"Percentual de duplicatas excedentes: {percentual:.2f}%")

if duplicadas > 0:
    print("\nForam encontradas linhas duplicadas.")

    # Exibe uma amostra das linhas duplicadas
    display(df[df.duplicated(keep=False)].head(10))
else:
    print("\nNenhuma linha duplicada encontrada.")
```

```python
import pandas as pd
from pathlib import Path

# Define o arquivo original
arquivo = Path("04_PENSE2024_SEM_NAN.parquet")

# Define o nome do novo arquivo no mesmo diretório
novo_arquivo = arquivo.with_name("05_PENSE2024_SEM_NAN_SEM_DUPLICADOS.parquet")

# Carrega o DataFrame original
df = pd.read_parquet(arquivo)

# Registra a quantidade original de linhas
linhas_antes = len(df)

# Remove linhas duplicadas, mantendo a primeira ocorrência
df_limpo = df.drop_duplicates(keep="first").copy()

# Registra a quantidade final de linhas
linhas_depois = len(df_limpo)

# Calcula quantas linhas foram removidas
removidas = linhas_antes - linhas_depois

# Salva o novo arquivo Parquet
df_limpo.to_parquet(novo_arquivo, index=False)

# Exibe o resumo
print(f"Linhas antes da remoção: {linhas_antes:,}")
print(f"Linhas após a remoção: {linhas_depois:,}")
print(f"Linhas removidas: {removidas:,}")
print(f"\nNovo arquivo salvo em: {novo_arquivo.resolve()}")
```

```python
import pandas as pd
from IPython.display import HTML, display

# 1. Carregar o dicionário de dados (se ainda não estiver carregado)
dicionario = pd.read_excel("00_Dicionário_Microdados_PENSE2024.xlsx")

# Criar mapeamento de Descrição -> Categoria (rótulos)
dico_map = {}
for _, row in dicionario.iterrows():
    desc = str(row["Descrição da Variável"]).strip()
    cat = str(row["Categoria"]) if pd.notna(row["Categoria"]) else ""
    cat_dict = {}
    if cat:
        for linha in cat.split("\n"):
            if " - " in linha:
                partes = linha.split(" - ", 1)
                cat_dict[partes[0].strip()] = partes[1].strip()
    dico_map[desc] = cat_dict

# 2. Gerar a tabela resumo para as colunas do df_limpo
tabela_dados = []

for col in df_limpo.columns:
    # Contar frequência e percentual
    counts = df_limpo[col].astype(str).str.strip().value_counts(dropna=False)
    total = len(df_limpo)
    
    # Mapear rótulos da categoria
    cat_dict = dico_map.get(col, {})
    
    respostas_formatadas = []
    for cod, qtd in counts.items():
        rotulo = cat_dict.get(cod, "Significado não encontrado" if cod == "-1" else "")
        pct = (qtd / total) * 100
        if rotulo:
            respostas_formatadas.append(f"[{cod}] {rotulo}: {pct:.2f}%")
        else:
            respostas_formatadas.append(f"[{cod}]: {pct:.2f}%")
    
    distribuicao_str = "\n".join(respostas_formatadas)
    tabela_dados.append({
        "Pergunta / Variável": col,
        "Distribuição das Respostas": distribuicao_str
    })

# 3. Criar DataFrame da tabela
df_tabela = pd.DataFrame(tabela_dados)

# 4. Estilizar com CSS (exatamente como na imagem)
styler = df_tabela.style.set_properties(**{
    'white-space': 'pre-wrap',
    'text-align': 'left'
})

print("========== EXIBIÇÃO DOS RESULTADOS NO IPYNB ==========\n")
display(HTML(styler.to_html()))
```
