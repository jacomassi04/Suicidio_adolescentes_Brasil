```python
import pandas as pd

arquivo = "07_PENSE2024_IDADE_13_A_17_ANOS.parquet"

df = pd.read_parquet(arquivo)

codigos_especiais = ["-9" , "-2" , "-1"]

# Verificar a ocorrência dos códigos
mascara = df.isin(codigos_especiais)

# Quantidade de ocorrências de cada código
print("Quantidade de -9:", mascara.eq(True).where(
    df.eq(-9), False
).sum().sum())

print("Quantidade de -2:", df.eq(-2).sum().sum())
print("Quantidade de -1:", df.eq(-1).sum().sum())

# Quantidade de registros únicos afetados
registros_afetados = mascara.any(axis=1).sum()

print("Total de registros:", len(df))
print("Registros com pelo menos um código especial:", registros_afetados)

print(
    "Percentual de registros afetados:",
    round(registros_afetados / len(df) * 100, 2),
    "%"
)
```

```python
import pandas as pd

arquivo = "01_PENSE2024_MICRODADOS.parquet"

df = pd.read_parquet(arquivo)

codigos_especiais = [-9, -2, -1]

# Identificar colunas com pelo menos uma ocorrência
mascara = df.isin(codigos_especiais)

colunas_afetadas = mascara.any(axis=0)

print("Total de colunas:", df.shape[1])

print(
    "Colunas com pelo menos um -9, -2 ou -1:",
    colunas_afetadas.sum()
)

print("\nQuantidade de colunas por código:")
print("Com -9:", df.eq(-9).any(axis=0).sum())
print("Com -2:", df.eq(-2).any(axis=0).sum())
print("Com -1:", df.eq(-1).any(axis=0).sum())

print("\nCódigos das colunas afetadas:")
print(colunas_afetadas[colunas_afetadas].index.tolist())
```
